```python
import json
import re
import nltk

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
faqs = [
    {
        "question": "What are the college timings?",
        "answer": "College timings are from 9:00 AM to 4:00 PM."
    },
    {
        "question": "How can I apply for admission?",
        "answer": "You can apply for admission through the college admission portal."
    },
    {
        "question": "Is attendance compulsory?",
        "answer": "Yes, students must maintain the required attendance percentage."
    },
    {
        "question": "Where can I check exam results?",
        "answer": "You can check your exam results through the student portal."
    },
    {
        "question": "How can I contact the college?",
        "answer": "You can contact the college office during working hours."
    }
]

print("FAQs loaded successfully!")
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)

questions = [preprocess(faq["question"]) for faq in faqs]

print(questions)
```

```python
vectorizer = TfidfVectorizer()

faq_vectors = vectorizer.fit_transform(questions)

print("TF-IDF vectorization completed!")
```

```python
def chatbot(user_question):
    user_question_clean = preprocess(user_question)

    user_vector = vectorizer.transform([user_question_clean])

    similarities = cosine_similarity(user_vector, faq_vectors)

    best_match_index = similarities.argmax()
    best_score = similarities[0][best_match_index]

    if best_score < 0.2:
        return "Sorry, I couldn't find an answer to your question."

    return faqs[best_match_index]["answer"]
```

```python
question = "How do I get admission?"

answer = chatbot(question)

print("User:", question)
print("Bot:", answer)
```

```python
print("🤖 FAQ Chatbot")
print("Type exit to stop the chatbot.")

while True:
    user_question = input("\nYou: ").strip()

    if user_question.lower().strip("'\"") == "exit":
        print("Bot: Thank you! Goodbye! 👋")
        break

    response = chatbot(user_question)
    print("Bot:", response)
```
